```python
import rdkit
from rdkit import Chem
import pandas as pd
```

```python
def convert_to_canonical_smiles(smiles: str) -> str:
    """
    Convert a given SMILES notation to its canonical SMILES representation using RDKit.

    Parameters:
        smiles (str): The input SMILES string.

    Returns:
        str: The canonical SMILES string or an error message if the input is invalid.
    """
    
    mol = Chem.MolFromSmiles(smiles)  # Convert SMILES to RDKit molecule object
    if mol:
        return Chem.MolToSmiles(mol, canonical=True)  # Convert back to canonical SMILES
    else:
        return "Invalid SMILES"
```

```python
df = pd.read_excel('12k_pfas.xlsx', sheet_name=1, index_col=0)
```

```python
df['SMILES']
```

```python
canons = []
for smi in df['SMILES']:
    try:
        canonized = convert_to_canonical_smiles(smiles=smi)
        #print(f'original: {smi}, canonized: {canonized}')
        canons.append(canonized)
    except:
        canons.append('failed')
    
df['canonical_SMILES'] = canons
```

```python
success = df[df['canonical_SMILES']!='failed']
```

```python
success.to_csv('12k_pfas_canonical.csv', sep=',')
```
